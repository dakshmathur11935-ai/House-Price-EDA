```python
Q1 = df['GrLivArea'].quantile(0.25)
Q3 = df['GrLivArea'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[
    (df['GrLivArea'] < lower) |
    (df['GrLivArea'] > upper)
]

print("IQR Outliers:", len(outliers))
```

```python
def iqr_outliers(column):

    Q1 = column.quantile(0.25)
    Q3 = column.quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    return df[(column < lower) | (column > upper)]
```

```python
def zscore_outliers(column):

    z = (column - column.mean()) / column.std()

    return df[abs(z) > 3]
```

```python
for column in ['GrLivArea', 'SalePrice']:

    iqr = iqr_outliers(df[column])
    zscore = zscore_outliers(df[column])

    print("\nColumn:", column)

    print("IQR Outliers:", len(iqr))

    print("Z-Score Outliers:", len(zscore))

    print("Common Outliers:", len(set(iqr.index) & set(zscore.index)))
```
